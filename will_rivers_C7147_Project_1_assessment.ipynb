## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
# Allowed nucleotide characters in a valid read
VALID_BASES = {"A", "C", "G", "T"}


def validate_read(sequence, reference):
    """Upper-case a read and check it matches the reference length and contains only A/C/G/T.

    Returns a dict with the upper-case sequence, whether it is valid, and a list of
    error codes ("wrong_length" then "invalid_characters").
    """

    #read needs to be normalised so lowercase is also accepted
    normalised_sequence = sequence.upper()
    errors = []

    #checks length first so the error codes always come out in the required order
    if len(normalised_sequence) != len(reference):
        errors.append("wrong_length")

    #report invalid characters only once, however many bad characters there are
    if any(base not in VALID_BASES for base in normalised_sequence):
        errors.append("invalid_characters")

    return {
        "sequence": normalised_sequence,
        "valid": len(errors) == 0,
        "errors": errors,
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python

def find_mutations(sequence, reference):
    """Compare a valid read to the reference and return a list of mismatches.

    Each mismatch is a dict with the position (counting from 1), the reference base
    and the read base. Returns an empty list if there are no mismatches.
    """
    mutations = []

    #loop over every index in the read (python counts from 0)
    for i in range(len(sequence)):
        read_base = sequence[i]
        reference_base = reference[i]

        #record only positions where the read differs from the reference
        if read_base != reference_base:
            mutations.append({
                "position": i + 1,  #this is what reads from position 1 (not 0)
                "reference_base": reference_base,
                "read_base": read_base,
            })

    return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Validate every read and find mismatches in the valid ones, without changing the input.

    Returns a new list with one dict per read, in the same order, adding the
    upper-case sequence, status ("valid"/"invalid"), errors and mutations.
    """

    analysed_reads = []

    for read in reads:
        #check the read is usable and get its upper-case sequence
        validation = validate_read(read["sequence"], reference)

        #only look for mismatches in valid reads; invalid reads get an empty list
        if validation["valid"]:
            status = "valid"
            mutations = find_mutations(validation["sequence"], reference)
        else:
            status = "invalid"
            mutations = []

        #build a new dictionary so the original read is left unchanged
        analysed_reads.append({
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": validation["errors"],
            "mutations": mutations,
        })

    return analysed_reads
````

### 💻 Enter your code here

````python
# Write your code here.

def calculate_percentage(count, total):
    """Return count as a percentage of total, or None if total is 0."""
    #avoid dividing by zero when a sample has no valid reads
    if total == 0:
        return None
    return count / total * 100


def get_top_positions(position_counts, n=3):
    """Return the n positions with the most mismatch events, earlier position winning ties."""
    #sort by highest count first (-count), then by lowest position to break ties
    ranked = sorted(position_counts.items(), key=lambda item: (-item[1], item[0]))
    return [position for position, count in ranked[:n]]


def count_sample_results(analysed_reads):
    """Count valid/invalid reads, mismatches, substitutions and positions for each sample.

    Returns a dict of counts per sample, with samples in the order first seen.
    """
    sample_counts = {}

    for read in analysed_reads:
        sample = read["sample"]

        #create an empty record the first time a sample is seen
        if sample not in sample_counts:
            sample_counts[sample] = {
                "group": read["group"],
                "valid_reads": 0,
                "invalid_reads": 0,
                "mismatch_reads": 0,
                "mismatch_events": 0,
                "substitution_counts": {},
                "position_counts": {},
            }
        counts = sample_counts[sample]

        #invalid reads are counted but contribute nothing else
        if read["status"] != "valid":
            counts["invalid_reads"] += 1
            continue

        counts["valid_reads"] += 1

        #a read with any mismatches counts once here, however many it has
        if read["mutations"]:
            counts["mismatch_reads"] += 1

        #every individual mismatch counts as one event
        for mutation in read["mutations"]:
            counts["mismatch_events"] += 1

            substitution = f"{mutation['reference_base']}>{mutation['read_base']}"
            counts["substitution_counts"][substitution] = counts["substitution_counts"].get(substitution, 0) + 1

            position = mutation["position"]
            counts["position_counts"][position] = counts["position_counts"].get(position, 0) + 1

    return sample_counts


def format_percentage(value, decimals=1):
    """Format a percentage to the given decimal places (default 1), or 'n/a' if it is None."""
    if value is None:
        return "n/a"
    return f"{value:.{decimals}f}%"

def group_samples(sample_summary):
    """Return a dict mapping each group name to a list of its sample names."""
    groups = {}
    for sample, summary in sample_summary.items():
        #setdefault creates an empty list the first time a group is seen
        groups.setdefault(summary["group"], []).append(sample)
    return groups


def print_summary_table(samples, sample_summary):
    """Print one row per sample with read counts, % with mismatch, events and top positions."""
    #make the sample column wide enough for the longest sample name
    width = max([len("Sample")] + [len(sample) for sample in samples])

    header = (f"{'Sample':<{width}}  {'Valid':>6}  {'Invalid':>7}  "
              f"{'% with mismatch':>15}  {'Events':>6}  Top positions")
    print(header)
    print("-" * len(header))

    for sample in samples:
        summary = sample_summary[sample]
        top_positions = ", ".join(str(p) for p in summary["top_positions"]) or "none"
        print(f"{sample:<{width}}  {summary['valid_reads']:>6}  {summary['invalid_reads']:>7}  "
              f"{format_percentage(summary['mismatch_percentage']):>15}  "
              f"{summary['mismatch_events']:>6}  {top_positions}")


def print_substitution_table(samples, sample_summary):
    """Print substitutions on rows and samples on columns, as count (% per 100 valid reads)."""
    #collect every substitution seen in any sample of this group, in alphabetical order
    substitutions = sorted({substitution
                            for sample in samples
                            for substitution in sample_summary[sample]["substitution_counts"]})
    if not substitutions:
        print("No substitutions observed.")
        return

    column_width = max([14] + [len(sample) for sample in samples])
    header = f"{'Substitution':<12}" + "".join(f"  {sample:>{column_width}}" for sample in samples)
    print(header)
    print("-" * len(header))

    for substitution in substitutions:
        row = f"{substitution.replace('>', ' → '):<12}"
        for sample in samples:
            summary = sample_summary[sample]
            count = summary["substitution_counts"].get(substitution, 0)
            percentage = calculate_percentage(count, summary["valid_reads"])
            cell = f"{count} ({format_percentage(percentage)})"
            row += f"  {cell:>{column_width}}"
        print(row)


def print_group_summaries(sample_summary):
    """Print a summary table and a substitution table for each group."""
    if not sample_summary:
        print("No reads to summarise.")
        return

    for group, samples in group_samples(sample_summary).items():
        print(f"\n===== Group: {group} =====")
        print("\nRead summary")
        print_summary_table(samples, sample_summary)
        print("\nSubstitutions: count (% per 100 valid reads)")
        print_substitution_table(samples, sample_summary)

        #clearly flag samples whose percentages could not be calculated
        for sample in samples:
            if sample_summary[sample]["valid_reads"] == 0:
                print(f"Note: {sample} has no valid reads, so its percentages cannot be calculated.")


def summarise_samples(analysed_reads):
    """Summarise mismatch patterns per sample from analyse_reads output and print tables per group.

    Percentages use valid reads only, and are None if a sample has no valid reads.
    Returns a dict with one summary dict per sample, or {} for empty input.
    """
    sample_summary = count_sample_results(analysed_reads)

    #add the values that are calculated from the counts
    for summary in sample_summary.values():
        summary["mismatch_percentage"] = calculate_percentage(summary["mismatch_reads"], summary["valid_reads"])
        summary["top_positions"] = get_top_positions(summary["position_counts"])

    print_group_summaries(sample_summary)
    return sample_summary
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).

🛠 **Parameters:**

- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.

def combine_position_counts(sample_summary, groups):
    """Sum the mismatch counts at every position across all samples in the given groups.

    Returns a list where index 0 holds position 1. The list runs up to the furthest
    position with a mismatch, so it is the read length when the last base is mutated.
    """
    #only use samples from the groups being compared
    selected = [summary for summary in sample_summary.values() if summary["group"] in groups]

    #make the list long enough for the furthest position with a mismatch
    max_position = max([0] + [position
                              for summary in selected
                              for position in summary["position_counts"]])
    combined_counts = [0] * max_position

    for summary in selected:
        for position, count in summary["position_counts"].items():
            combined_counts[position - 1] += count  #positions count from 1, list index from 0

    return combined_counts


def select_top_positions(combined_counts, n=3):
    """Return the n positions with the highest combined counts, earlier position winning ties."""
    #turn the list back into {position: count}, skipping positions with no mismatches
    position_counts = {index + 1: count for index, count in enumerate(combined_counts) if count > 0}
    #reuse the Step 4 helper so ties are broken the same way
    return get_top_positions(position_counts, n)

def find_excluded_samples(sample_summary, groups):
    """Return the samples in the given groups that have no valid reads."""
    return [sample for sample, summary in sample_summary.items()
            if summary["group"] in groups and summary["valid_reads"] == 0]



def get_position_percentages(sample_summary, group, position):
    """Return one mismatch percentage per sample in a group at a single position.

    Samples with no valid reads are skipped, as their percentage cannot be calculated.
    """
    percentages = []
    for summary in sample_summary.values():
        if summary["group"] != group:
            continue

        #mismatch events at one position = reads with a mismatch there (one base per read)
        mismatch_reads = summary["position_counts"].get(position, 0)
        percentage = calculate_percentage(mismatch_reads, summary["valid_reads"])

        #calculate_percentage gives None when there are no valid reads
        if percentage is not None:
            percentages.append(percentage)

    return percentages

def calculate_mean(values):
    """Return the mean of a list of numbers, or None if the list is empty."""
    if not values:
        return None
    return sum(values) / len(values)

def calculate_welch_df(control_percentages, treated_percentages):
    """Return the Welch-Satterthwaite degrees of freedom, or None if they cannot be calculated."""
    control_n = len(control_percentages)
    treated_n = len(treated_percentages)
    if control_n < 2 or treated_n < 2:
        return None

    #squared standard error of each group's mean (ddof=1 gives the sample variance)
    control_se2 = np.var(control_percentages, ddof=1) / control_n
    treated_se2 = np.var(treated_percentages, ddof=1) / treated_n

    denominator = control_se2 ** 2 / (control_n - 1) + treated_se2 ** 2 / (treated_n - 1)
    #no variation in either group means df cannot be estimated
    if denominator == 0:
        return None
    return float((control_se2 + treated_se2) ** 2 / denominator)

def format_p_value(p_value):
    """Format a p-value in scientific style, e.g. 'p = 0.012' or 'p < 0.001'."""
    if p_value is None:
        return "p = n/a"
    if p_value < 0.001:
        return "p < 0.001"
    return f"p = {p_value:.3f}"

def print_position_result(result, df, control_group, treated_group):
    """Print one line reporting group means, difference, sample numbers and Welch's t-test."""
    if result["difference"] is None:
        difference = "n/a"
    else:
        difference = f"{result['difference']:+.2f} percentage points"

    #only report t and df when the test could be run
    if result["statistic"] is None:
        test = "Welch's t-test could not be calculated"
    else:
        df_text = f"{df:.2f}" if df is not None else "n/a"
        test = f"Welch's t({df_text}) = {result['statistic']:.2f}, {format_p_value(result['p_value'])}"

    print(f"Position {result['position']}: "
          f"{treated_group} {format_percentage(result['treated_mean'], 2)} (n = {result['treated_n']}) vs "
          f"{control_group} {format_percentage(result['control_mean'], 2)} (n = {result['control_n']}); "
          f"difference = {difference}; {test}")

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Compare mismatch percentages between two groups at the top 3 combined mismatch positions.

    Uses position_counts from summarise_samples, picks the top positions across both groups,
    then runs a two-sided Welch's t-test on per-sample percentages at each position.
    Returns a list with one result dict per position, using None where values cannot be calculated.
    """
    groups = [control_group, treated_group]

    #samples with no valid reads cannot give a percentage, so report and leave them out
    for sample in find_excluded_samples(sample_summary, groups):
        print(f"Note: {sample} has no valid reads and is excluded from the comparison.")

    combined_counts = combine_position_counts(sample_summary, groups)
    top_positions = select_top_positions(combined_counts)

    if not top_positions:
        print(f"No mismatches found in {control_group} or {treated_group} samples to compare.")
        return []

    print(f"Top mismatch positions: {treated_group} vs {control_group} "
          f"(mean % of valid reads with a mismatch, two-sided Welch's t-test)")

    results = []
    for position in top_positions:
        control_percentages = get_position_percentages(sample_summary, control_group, position)
        treated_percentages = get_position_percentages(sample_summary, treated_group, position)

        control_mean = calculate_mean(control_percentages)
        treated_mean = calculate_mean(treated_percentages)

        #difference is treated minus control, only if both means exist
        if control_mean is None or treated_mean is None:
            difference = None
        else:
            difference = treated_mean - control_mean

        statistical_result = calculate_statistics(control_percentages, treated_percentages)

        result = {
            "position": position,
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistical_result["statistic"],
            "p_value": statistical_result["p_value"],
        }
        results.append(result)

        #df is only needed for printing, so it is not stored in the result
        df = calculate_welch_df(control_percentages, treated_percentages)
        print_position_result(result, df, control_group, treated_group)

    return results


````
